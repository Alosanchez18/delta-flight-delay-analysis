```python
import pandas as pd
import matplotlib.pyplot as plt # the charting tool

df_airports = pd.read_csv("outputs/query1_airport_delays.csv")
df_monthly = pd.read_csv("outputs/query3_monthly_trend.csv")

print("Loaded.")

# Take the 10 worst airports, smallest-to-largest so the bar chart reads top-down
top10 = df_airports.head(10).sort_values("delay_rate_pct")
plt.figure(figsize=(10,6))
plt.barh(top10["airport"],top10["delay_rate_pct"],color="#9E2A2B") # Delta red
plt.xlabel("Delay Rate (%)")
plt.title("Top 10 Delta Airports by Delay Rate (2023-2025)")
plt.tight_layout()
plt.savefig("outputs/chart1_airport_delays.png",dpi=150) #save it for repo
plt.show()

```

```python
plt.figure(figsize=(11,5))
plt.plot(df_monthly["month"], df_monthly["delay_rate_pct"],
         color="#9E2A2B", linewidth=2.5, marker="o")
plt.xticks(range(1,13))
plt.xlabel("Month (1 = Jan ... 12 = Dec)")
plt.ylabel("Delay Rate (%)")
plt.title("Delta Delay Rate by Month (2023-2025)")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig("outputs/chart2_monthly_trend.png", dpi=150)
plt.show()
```

```python
worst_airport = df_airports.iloc[0] #first row = highest delay rate
worst_month = df_monthly.loc[df_monthly["delay_rate_pct"].idxmax()]

print("KEY FINDINGS - Delta Flight Delays (2023-2025)")
print("-" * 45)
print(f"Worst airport: {worst_airport['airport_name']}"
      f"({worst_airport['delay_rate_pct']}% of arrivals delayed)")
print(f"Worst month: month {int(worst_month['month'])}"
      f"({worst_month['delay_rate_pct']}%)")
```
